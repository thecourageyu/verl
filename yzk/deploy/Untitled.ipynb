```python
import zmq
import numpy as np
import time

context = zmq.Context()
socket = context.socket(zmq.PUB)
socket.connect("tcp://127.0.0.1:5555")

# 模擬生成一個平滑變化的手臂/軀幹目標軌跡
t = 0.0
while True:
    ref_pose = np.zeros(18, dtype=np.float32)
    # 讓某個關節或目標位置隨時間做平滑的正弦擺動
    ref_pose[0] = 0.2 * np.sin(2 * np.pi * 0.5 * t)  
    
    socket.send(ref_pose.tobytes())
    t += 0.02  # 50 Hz
    time.sleep(0.02)
```
